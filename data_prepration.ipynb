```python
import pandas as pd

# --- Configuration ---
input_file = "PaidClaimsInsertedDate01012023_05012023.xlsx"
output_file = "sampled_data.csv"
num_rows = 19901  # number of rows to extract

# --- Read Excel file ---
print(" Reading Excel file...")
df = pd.read_excel(input_file)

# --- Shuffle and pick 19,901 rows ---
print(" Shuffling and sampling rows...")
sampled_df = df.sample(n=num_rows, random_state=42)

# --- Save to CSV ---
sampled_df.to_csv(output_file, index=False)

print(f" {num_rows} shuffled rows successfully saved to '{output_file}'")
```

```python
import pandas as pd

# --- File names ---
input_file = "docs_and_raw_data/paid-claims.csv"
output_file = "paid_renamed.csv"

# --- Read the CSV file ---
df = pd.read_csv(input_file)

# --- Rename columns (Paid#_Status -> Paid#) ---
rename_map = {
    f"Paid{i}_Status": f"Paid{i}"
    for i in range(1, 7)
}

df.rename(columns=rename_map, inplace=True)

# --- Save the modified CSV ---
df.to_csv(output_file, index=False)

print(f" Columns renamed successfully and saved as '{output_file}'")
```

```python
import pandas as pd

# --- File paths ---
input_file = "docs/NotPaidClaimsInsertedDate01012023_05012023.xlsx"
output_file = "unpaid_claims.csv"

# --- Read Excel file ---
df = pd.read_excel(input_file)

# --- Save as CSV ---
df.to_csv(output_file, index=False)

print(f" Excel file '{input_file}' successfully converted to CSV as '{output_file}'")
```

```python
import pandas as pd

# --- Step 1: Read both CSV files ---
paid_csv = "data/new-model-traing-data/paid_renamed.csv"
unpaid_csv = "data/new-model-traing-data/unpaid_claims.csv"

paid_df = pd.read_csv(paid_csv)
unpaid_df = pd.read_csv(unpaid_csv)

# --- Step 2: Add numeric label column for model training ---
paid_df["Claim_status"] = 1   # Paid = 1
unpaid_df["Claim_status"] = 0 # Unpaid = 0

# --- Step 3: Concatenate both datasets ---
merged_df = pd.concat([paid_df, unpaid_df], ignore_index=True, sort=False)

# --- Step 4: Save merged CSV ---
merged_df.to_csv("merged_claims.csv", index=False)

print(f"Merged CSV created successfully with shape: {merged_df.shape}")
print("Missing columns automatically filled with NaN where not available.")
print("Label column 'Claim_status' added (1 = Paid, 0 = Unpaid).")
```
